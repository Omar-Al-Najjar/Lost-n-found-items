````python
# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 5-A  Image loader  (URL â†’ PIL Image)
# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
def load_image_from_url(url: str) -> Image.Image:
    """Download and return a PIL image from a public URL."""
    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept": "image/avif,image/webp,image/apng,image/*,*/*;q=0.8",
    }
    response = requests.get(url, headers=headers, timeout=20)
    response.raise_for_status()
    return Image.open(BytesIO(response.content)).convert("RGB")


def display_image(img: Image.Image, caption: str = ""):
    """Resize for notebook display and show."""
    import matplotlib.pyplot as plt
    fig, ax = plt.subplots(figsize=(5, 4))
    ax.imshow(img)
    ax.axis("off")
    if caption:
        ax.set_title(caption, fontsize=10, wrap=True)
    plt.tight_layout()
    plt.show()


def pil_to_base64(img: Image.Image, image_format: str = "JPEG") -> str:
    """Convert a PIL image to a base64 data URI for Kimi vision calls."""
    buffer = BytesIO()
    fmt = (image_format or "JPEG").upper()
    if fmt == "JPG":
        fmt = "JPEG"
    img.save(buffer, format=fmt)
    mime = "image/jpeg" if fmt == "JPEG" else f"image/{fmt.lower()}"
    b64 = base64.b64encode(buffer.getvalue()).decode("utf-8")
    return f"data:{mime};base64,{b64}"


print("âœ… Image helpers defined (load_image_from_url, display_image, pil_to_base64).")
````

````python

# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 5-B  PIPELINE 1 â€” Vision â†’ Structured Found-Item Description
# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
FOUND_ITEM_SYSTEM_PROMPT = """You are an AI assistant for a Lost & Found application.
A user found an item and provided a photo plus a short raw description.

Return ONLY one valid JSON object with exactly these fields:
{
  "ai_description": "<2-4 sentence grounded description of only what is visible or strongly implied>",
  "item_type": "<specific item type like wallet, backpack, earbuds case, car keys, sunglasses, phone, document folder, etc.>",
  "category": "<one of: Electronics | Clothing | Jewellery | Bag/Wallet | Keys | Documents | Pet | Toy | Sports | Other>",
  "color": "<dominant color(s)>",
  "material": "<material if visible or strongly suggested, else Unknown>",
  "brand": "<brand/make if visible, else Unknown>",
  "visible_contents": ["<cards>", "<cash>", "<keys>", "..."],
  "distinctive_features": ["<zipper>", "<logo>", "<sticker>", "<keychain>", "<scratches>", "..."],
  "condition": "<New | Good | Used | Damaged | Unknown>",
  "attribute_confidence": "<High | Medium | Low>",
  "search_keywords": ["<keyword1>", "<keyword2>", "<keyword3>", "<keyword4>", "<keyword5>", "<keyword6>"]
}

Rules:
- Use the photo as the main source of truth.
- Use the user description only as supporting context.
- Never invent hidden details.
- If a field is unknown, use "Unknown".
- visible_contents and distinctive_features must be arrays.
- search_keywords must be short, lowercase-friendly retrieval terms.
- Keep ai_description factual and concise.
- Output JSON only. No markdown fences, no extra commentary.
"""

LOST_ITEM_PARSE_PROMPT = """You are helping a Lost & Found search engine.
Convert the user's lost-item description into structured JSON.

Return ONLY one valid JSON object with exactly these fields:
{
  "query_text": "<cleaned version of the user description>",
  "item_type": "<specific item type or Unknown>",
  "category": "<one of: Electronics | Clothing | Jewellery | Bag/Wallet | Keys | Documents | Pet | Toy | Sports | Other>",
  "color": "<dominant color(s) or Unknown>",
  "material": "<material or Unknown>",
  "brand": "<brand or Unknown>",
  "likely_contents": ["<cards>", "<cash>", "..."],
  "distinctive_features": ["<logo>", "<zipper>", "<engraving>", "..."],
  "location_clues": ["<university>", "<gate>", "..."],
  "time_clues": ["<this morning>", "<yesterday>", "..."],
  "search_keywords": ["<keyword1>", "<keyword2>", "<keyword3>", "<keyword4>", "<keyword5>", "<keyword6>"]
}

Rules:
- Infer cautiously.
- Do not hallucinate a brand, color, or material.
- Normalize vague descriptions into short retrieval-friendly terms.
- Arrays must always be arrays.
- Output JSON only.
"""

def _extract_json_object(text: str) -> dict:
    """Parse a JSON object even if the model wraps it with extra text."""
    text = (text or "").strip()
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        match = re.search(r"\{.*\}", text, flags=re.DOTALL)
        if match:
            return json.loads(match.group(0))
        raise

def analyze_found_item(image, user_description):
    image_data_uri = pil_to_base64(image)

    response = kimi_client.chat.completions.create(
        model=KIMI_MODEL,
        messages=[
            {"role": "system", "content": FOUND_ITEM_SYSTEM_PROMPT},
            {
                "role": "user",
                "content": [
                    {"type": "image_url", "image_url": {"url": image_data_uri}},
                    {
                        "type": "text",
                        "text": (
                            f"Finder description: {user_description}\n\n"
                            "Analyze the image and return only the JSON object."
                        ),
                    },
                ],
            },
        ],
        response_format={"type": "json_object"},
        max_tokens=1000,
        extra_body={"thinking": {"type": "disabled"}},
    )

    content = response.choices[0].message.content
    return normalize_ai_result(_extract_json_object(content))

def _fallback_parse_lost_item_description(lost_description: str) -> dict:
    text = (lost_description or "").strip()
    low = text.lower()

    item_type = "Unknown"
    if "wallet" in low:
        item_type = "wallet"
    elif "backpack" in low or "bag" in low:
        item_type = "backpack" if "backpack" in low else "bag"
    elif "airpods" in low or "earbuds" in low:
        item_type = "earbuds case" if "case" in low else "earbuds"
    elif "keys" in low or "keychain" in low:
        item_type = "keys"
    elif "sunglasses" in low or "glasses" in low:
        item_type = "sunglasses"
    elif "phone" in low or "iphone" in low:
        item_type = "phone"

    color = "Unknown"
    for candidate in ["black", "brown", "white", "silver", "gold", "blue", "red", "green", "pink", "gray", "grey", "yellow"]:
        if candidate in low:
            color = candidate
            break

    material = "Unknown"
    for candidate in ["leather", "metal", "plastic", "fabric", "canvas"]:
        if candidate in low:
            material = candidate
            break

    brand = "Unknown"
    for candidate in ["apple", "samsung", "nike", "adidas", "ray-ban", "rayban"]:
        if candidate in low:
            brand = candidate
            break

    category = infer_category(item_type=item_type, raw_category="Other", text=text)

    likely_contents = [x for x in ["cards", "cash", "id", "id cards", "license"] if x in low]
    distinctive_features = [x for x in ["zipper", "logo", "sticker", "engraving", "keychain"] if x in low]
    location_clues = [x for x in ["university", "gate", "park", "mall", "entrance", "library"] if x in low]
    time_clues = [x for x in ["this morning", "morning", "today", "yesterday", "afternoon", "evening"] if x in low]

    keywords = list(dict.fromkeys(
        [item_type, category.lower(), color, material, brand] + likely_contents + distinctive_features + location_clues
    ))
    keywords = [x for x in keywords if x and x != "Unknown"]

    return {
        "query_text": text,
        "item_type": item_type,
        "category": category,
        "color": color,
        "material": material,
        "brand": brand,
        "likely_contents": likely_contents,
        "distinctive_features": distinctive_features,
        "location_clues": location_clues,
        "time_clues": time_clues,
        "search_keywords": keywords[:8],
    }

def parse_lost_item_description(lost_description: str) -> dict:
    try:
        response = kimi_client.chat.completions.create(
            model=KIMI_MODEL,
            messages=[
                {"role": "system", "content": LOST_ITEM_PARSE_PROMPT},
                {"role": "user", "content": lost_description},
            ],
            response_format={"type": "json_object"},
            max_tokens=700,
            extra_body={"thinking": {"type": "disabled"}},
        )
        parsed = _extract_json_object(response.choices[0].message.content)
    except Exception:
        parsed = _fallback_parse_lost_item_description(lost_description)

    parsed = parsed or {}
    fallback = _fallback_parse_lost_item_description(lost_description)

    normalized = {
        "query_text": str(parsed.get("query_text") or fallback["query_text"]).strip(),
        "item_type": str(parsed.get("item_type") or fallback["item_type"]).strip() or "Unknown",
        "category": infer_category(
            item_type=str(parsed.get("item_type") or fallback["item_type"]),
            raw_category=str(parsed.get("category") or fallback["category"]),
            text=lost_description,
        ),
        "color": str(parsed.get("color") or fallback["color"]).strip() or "Unknown",
        "material": str(parsed.get("material") or fallback["material"]).strip() or "Unknown",
        "brand": str(parsed.get("brand") or fallback["brand"]).strip() or "Unknown",
        "likely_contents": ensure_list(parsed.get("likely_contents") or fallback["likely_contents"]),
        "distinctive_features": ensure_list(parsed.get("distinctive_features") or fallback["distinctive_features"]),
        "location_clues": ensure_list(parsed.get("location_clues") or fallback["location_clues"]),
        "time_clues": ensure_list(parsed.get("time_clues") or fallback["time_clues"]),
        "search_keywords": ensure_list(parsed.get("search_keywords") or fallback["search_keywords"]),
    }

    normalized["search_keywords"] = expand_search_keywords(normalized)
    return normalized


print("âœ… Pipeline 1 helper defined  (analyze_found_item).")
print("âœ… Lost-item parser defined  (parse_lost_item_description).")


# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 5-C  PIPELINE 2 â€” Lost-Item Query â†’ Match Explanation
# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
MATCH_EXPLANATION_PROMPT = """You are an AI assistant for a Lost & Found application.
Write a short, grounded explanation for a possible match.

Lost query:
{lost_description}

Structured lost query:
{lost_structured}

Found item record:
{found_structured}

Evidence supporting the match:
{evidence_lines}

Potential contradictions or uncertainty:
{contradiction_lines}

Ranking score: {score:.2f}
Confidence tier: {confidence_label}

Rules:
- Mention only evidence provided above.
- Be concise and honest.
- End with 1 sentence on what the owner should verify with the finder.
"""

def explain_match(lost_description: str, found_item: dict, score: float) -> str:
    """Generate a grounded human-readable explanation for a candidate match."""
    evidence_lines = "\n".join(f"- {x}" for x in found_item.get("evidence", [])[:6]) or "- semantic similarity"
    contradiction_lines = "\n".join(f"- {x}" for x in found_item.get("contradictions", [])[:4]) or "- no major contradictions detected"

    lost_structured = json.dumps(found_item.get("lost_structured", {}), ensure_ascii=False)
    found_structured = json.dumps({
        k: found_item.get(k) for k in [
            "item_type", "category", "color", "material", "brand",
            "visible_contents", "distinctive_features", "condition", "location_found"
        ]
    }, ensure_ascii=False)

    prompt = MATCH_EXPLANATION_PROMPT.format(
        lost_description=lost_description,
        lost_structured=lost_structured,
        found_structured=found_structured,
        evidence_lines=evidence_lines,
        contradiction_lines=contradiction_lines,
        score=score,
        confidence_label=found_item.get("confidence_label", "Unknown"),
    )

    try:
        response = kimi_client.chat.completions.create(
            model=KIMI_MODEL,
            messages=[{"role": "user", "content": prompt}],
            max_tokens=220,
            extra_body={"thinking": {"type": "disabled"}},
        )
        return response.choices[0].message.content.strip()
    except Exception:
        summary = "; ".join(found_item.get("evidence", [])[:3]) or "The semantic description is reasonably close."
        contradiction = ""
        if found_item.get("contradictions"):
            contradiction = " There are some uncertainties: " + "; ".join(found_item["contradictions"][:2]) + "."
        return (
            f"This looks like a possible match because {summary.lower()}.{contradiction} "
            f"Ask the finder to verify the most distinctive details before claiming it."
        )

print("âœ… Match explanation helper defined  (explain_match).")
````

````python
import re


# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
# 5-D  Database write / search helpers
# â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
from datetime import datetime, UTC

ALLOWED_CATEGORIES = {
    "Electronics", "Clothing", "Jewellery", "Bag/Wallet", "Keys",
    "Documents", "Pet", "Toy", "Sports", "Other"
}

CATEGORY_HINTS = {
    "Bag/Wallet": {"wallet", "billfold", "card holder", "cardholder", "backpack", "bag", "purse", "handbag"},
    "Keys": {"keys", "key", "keychain", "car keys"},
    "Electronics": {"airpods", "earbuds", "earbud", "headphones", "phone", "iphone", "ipad", "tablet", "laptop", "charger", "electronics", "sunglasses"},
    "Documents": {"passport", "document", "id", "license", "folder", "papers"},
    "Clothing": {"hoodie", "jacket", "shirt", "coat", "cap", "hat", "scarf"},
    "Jewellery": {"ring", "bracelet", "necklace", "watch", "earring"},
    "Sports": {"ball", "racket", "helmet", "sports"},
}

CATEGORY_CONTRADICTIONS = {
    ("Bag/Wallet", "Keys"),
    ("Bag/Wallet", "Electronics"),
    ("Keys", "Electronics"),
    ("Documents", "Keys"),
}

# Item-type pairs that are clearly distinct even within the same category.
# If both sides have a known item_type in one of these pairs, apply an extra penalty.
ITEM_TYPE_CONTRADICTIONS = {
    ("wallet", "backpack"), ("wallet", "bag"), ("wallet", "purse"),
    ("backpack", "wallet"), ("bag", "wallet"), ("purse", "wallet"),
    ("airpods", "phone"), ("phone", "airpods"),
    ("keys", "wallet"), ("wallet", "keys"),
}

def ensure_list(value):
    if value is None:
        return []
    if isinstance(value, list):
        return [str(x).strip() for x in value if str(x).strip() and str(x).strip().lower() != "unknown"]
    if isinstance(value, str):
        parts = [x.strip() for x in re.split(r"[,;\n]+", value) if x.strip()]
        return [x for x in parts if x.lower() != "unknown"]
    return [str(value).strip()]

def _safe_join_keywords(value):
    return " ".join(ensure_list(value))

def normalize_token(text: str) -> str:
    return re.sub(r"[^a-z0-9]+", " ", str(text).lower()).strip()

def tokenize(text: str) -> list[str]:
    return [tok for tok in normalize_token(text).split() if tok]

def infer_category(item_type: str = "Unknown", raw_category: str = "Other", text: str = "") -> str:
    tokens = set(tokenize(" ".join([item_type or "", raw_category or "", text or ""])))
    for category, hints in CATEGORY_HINTS.items():
        if tokens & set(tokenize(" ".join(hints))):
            return category
    raw_category = str(raw_category or "Other").strip()
    return raw_category if raw_category in ALLOWED_CATEGORIES else "Other"

def expand_search_keywords(payload: dict) -> list[str]:
    """Build structured keyword list for attribute matching (NOT used in embeddings)."""
    base = []
    for key in [
        "item_type", "category", "color", "material", "brand",
        "visible_contents", "likely_contents", "distinctive_features",
        "location_clues", "search_keywords"
    ]:
        value = payload.get(key)
        if isinstance(value, list):
            base.extend(value)
        else:
            base.append(value)

    text = " ".join(str(x) for x in base if x)
    tokens = tokenize(text)

    synonyms = []
    joined = " ".join(tokens)
    if "wallet" in joined:
        synonyms += ["billfold", "card holder", "cardholder"]
    if "airpods" in joined or "earbuds" in joined:
        synonyms += ["earbuds", "earbud case", "airpods case", "wireless earbuds"]
    if "keys" in joined:
        synonyms += ["keychain", "car keys"]
    if "backpack" in joined:
        synonyms += ["bag", "school bag"]
    if "sunglasses" in joined:
        synonyms += ["glasses", "shades"]

    combined = tokens + tokenize(" ".join(synonyms))
    unique = []
    for tok in combined:
        if tok not in unique and tok not in {"unknown", "other"}:
            unique.append(tok)
    return unique[:18]

def normalize_ai_result(ai_result: dict) -> dict:
    """Map any slightly different AI schema into one canonical record shape."""
    ai_result = ai_result or {}

    item_type = str(ai_result.get("item_type") or ai_result.get("type") or "Unknown").strip() or "Unknown"
    category = infer_category(
        item_type=item_type,
        raw_category=str(ai_result.get("category", "Other")),
        text=json.dumps(ai_result, ensure_ascii=False),
    )
    color = str(ai_result.get("color", "Unknown")).strip() or "Unknown"
    material = str(ai_result.get("material", "Unknown")).strip() or "Unknown"
    brand = str(ai_result.get("brand", "Unknown")).strip() or "Unknown"
    condition = str(ai_result.get("condition", "Unknown")).strip() or "Unknown"
    attribute_confidence = str(ai_result.get("attribute_confidence", "Unknown")).strip() or "Unknown"

    visible_contents = ensure_list(ai_result.get("visible_contents"))
    distinctive_features = ensure_list(
        ai_result.get("distinctive_features")
        or ai_result.get("notable_features")
        or ai_result.get("features")
    )

    search_keywords = ensure_list(ai_result.get("search_keywords"))

    ai_description = ai_result.get("ai_description")
    if not ai_description:
        parts = [color, material, brand if brand != "Unknown" else "", item_type, category, condition]
        parts = [str(p).strip() for p in parts if str(p).strip() and str(p).strip() != "Unknown"]
        ai_description = " ".join(parts).strip() or "Found item"
        if distinctive_features:
            ai_description += ". Distinctive features: " + ", ".join(distinctive_features[:5])

    normalized = {
        "ai_description": str(ai_description).strip(),
        "item_type": item_type,
        "category": category,
        "color": color,
        "material": material,
        "brand": brand,
        "visible_contents": visible_contents,
        "distinctive_features": distinctive_features,
        "condition": condition,
        "attribute_confidence": attribute_confidence,
        "search_keywords": expand_search_keywords({
            "item_type": item_type,
            "category": category,
            "color": color,
            "material": material,
            "brand": brand,
            "visible_contents": visible_contents,
            "distinctive_features": distinctive_features,
            "search_keywords": search_keywords,
        }),
    }
    return normalized

# IMPROVEMENT 1: Lean embed texts -- no synonym expansion, no keyword bloat.
# Previously build_*_embed_text included expanded search_keywords, injecting synonym
# clouds (e.g. "billfold cardholder" for wallet, "wireless earbuds earbud case" for
# AirPods) that inflated cosine-similarity between unlike items.
# Synonyms/keywords are still used for structured attribute matching, just not for
# embedding.

def build_found_embed_text(normalized: dict, user_description: str, location_found: str) -> str:
    """Concise, fact-only text for embedding - no synonym expansion."""
    return " ".join([
        normalized["ai_description"],
        normalized["item_type"],
        normalized["category"],
        normalized["color"],
        normalized["material"],
        normalized["brand"],
        _safe_join_keywords(normalized["visible_contents"]),
        _safe_join_keywords(normalized["distinctive_features"]),
        str(user_description),
        str(location_found),
    ]).strip()

def build_lost_embed_text(parsed_lost: dict, raw_description: str) -> str:
    """Concise, fact-only text for embedding - no synonym expansion."""
    return " ".join([
        parsed_lost.get("query_text", raw_description),
        parsed_lost.get("item_type", ""),
        parsed_lost.get("category", ""),
        parsed_lost.get("color", ""),
        parsed_lost.get("material", ""),
        parsed_lost.get("brand", ""),
        _safe_join_keywords(parsed_lost.get("likely_contents")),
        _safe_join_keywords(parsed_lost.get("distinctive_features")),
        _safe_join_keywords(parsed_lost.get("location_clues")),
    ]).strip()

def overlap_score(a: list[str], b: list[str]) -> float:
    sa = set(tokenize(" ".join(a)))
    sb = set(tokenize(" ".join(b)))
    if not sa or not sb:
        return 0.0
    return len(sa & sb) / len(sa | sb)

def text_match(a: str, b: str) -> bool:
    ta = set(tokenize(a))
    tb = set(tokenize(b))
    return bool(ta and tb and (ta & tb))

# IMPROVEMENT 2: Stricter score_candidate
# Changes vs v1:
#   Semantic weight: 0.58 -> 0.50  (structured attributes now carry more weight)
#   Item-type mismatch penalty: 0.18 -> 0.28
#   Item-type contradiction surcharge (e.g. wallet vs backpack): extra -0.15
#   Category contradiction penalty: 0.20 -> 0.35
#   Brand mismatch penalty: 0.08 -> 0.22  (Samsung != Apple should near-disqualify)
#   Brand match bonus: 0.12 -> 0.18  (reward exact brand match)
#   High confidence: score >= 0.72 -> 0.75
#   Medium confidence: score >= 0.52 + 2 hits -> score >= 0.60 + 3 hits

def score_candidate(parsed_lost: dict, item: dict, semantic_score: float) -> dict:
    evidence = []
    contradictions = []
    bonus = 0.0
    penalty = 0.0

    # Item type
    lost_type  = parsed_lost.get("item_type", "").lower().strip()
    found_type = item.get("item_type", "").lower().strip()

    if text_match(lost_type, found_type):
        bonus += 0.22
        evidence.append(f"item type aligns ({parsed_lost.get('item_type')} <-> {item.get('item_type')})")
    elif lost_type not in {"", "unknown"} and found_type not in {"", "unknown"}:
        is_type_contradiction = (
            (lost_type, found_type) in ITEM_TYPE_CONTRADICTIONS or
            (found_type, lost_type) in ITEM_TYPE_CONTRADICTIONS
        )
        if is_type_contradiction:
            penalty += 0.43  # 0.28 base + 0.15 surcharge for known contradictory pair
            contradictions.append(f"item type contradiction ({lost_type} vs {found_type})")
        else:
            penalty += 0.28
            contradictions.append(f"item type differs ({parsed_lost.get('item_type')} vs {item.get('item_type')})")

    # Category
    lost_cat  = parsed_lost.get("category")
    found_cat = item.get("category")
    if lost_cat == found_cat and lost_cat != "Other":
        bonus += 0.15
        evidence.append(f"category matches ({found_cat})")
    elif (lost_cat, found_cat) in CATEGORY_CONTRADICTIONS or (found_cat, lost_cat) in CATEGORY_CONTRADICTIONS:
        penalty += 0.35
        contradictions.append(f"category conflict ({lost_cat} vs {found_cat})")
    elif lost_cat != found_cat and lost_cat not in {"Other", None} and found_cat not in {"Other", None}:
        penalty += 0.10
        contradictions.append(f"category mismatch ({lost_cat} vs {found_cat})")

    # Color
    if parsed_lost.get("color") not in {"", "Unknown"} and item.get("color") not in {"", "Unknown"}:
        if text_match(parsed_lost.get("color", ""), item.get("color", "")):
            bonus += 0.12
            evidence.append(f"color aligns ({parsed_lost.get('color')} <-> {item.get('color')})")
        else:
            penalty += 0.10
            contradictions.append(f"color mismatch ({parsed_lost.get('color')} vs {item.get('color')})")

    # Material
    if parsed_lost.get("material") not in {"", "Unknown"} and item.get("material") not in {"", "Unknown"}:
        if text_match(parsed_lost.get("material", ""), item.get("material", "")):
            bonus += 0.10
            evidence.append(f"material aligns ({parsed_lost.get('material')} <-> {item.get('material')})")
        else:
            penalty += 0.06
            contradictions.append(f"material mismatch ({parsed_lost.get('material')} vs {item.get('material')})")

    # Brand -- heavily weighted; a confirmed mismatch is near-disqualifying
    lost_brand  = parsed_lost.get("brand", "")
    found_brand = item.get("brand", "")
    if lost_brand not in {"", "Unknown"} and found_brand not in {"", "Unknown"}:
        if text_match(lost_brand, found_brand):
            bonus += 0.18   # raised from 0.12
            evidence.append(f"brand aligns ({lost_brand} <-> {found_brand})")
        else:
            penalty += 0.22  # raised from 0.08
            contradictions.append(f"brand mismatch ({lost_brand} vs {found_brand})")

    # Overlapping structured lists
    feature_overlap  = overlap_score(parsed_lost.get("distinctive_features", []), item.get("distinctive_features", []))
    contents_overlap = overlap_score(parsed_lost.get("likely_contents", []),      item.get("visible_contents", []))
    keyword_overlap  = overlap_score(parsed_lost.get("search_keywords", []),      item.get("search_keywords", []))
    location_overlap = overlap_score(parsed_lost.get("location_clues", []),       tokenize(item.get("location_found", "")))

    if feature_overlap > 0:
        bonus += min(0.14, feature_overlap * 0.20)
        evidence.append("distinctive features overlap")
    if contents_overlap > 0:
        bonus += min(0.10, contents_overlap * 0.18)
        evidence.append("contents/details overlap")
    if keyword_overlap > 0:
        bonus += min(0.08, keyword_overlap * 0.12)
        evidence.append("retrieval keywords overlap")
    if location_overlap > 0:
        bonus += min(0.05, location_overlap * 0.08)
        evidence.append("location context overlaps")

    # Semantic weight reduced from 0.58 to 0.50
    final_score = max(0.0, min(1.0, 0.50 * semantic_score + bonus - penalty))

    # Confidence tiers (tightened)
    # High:   score >= 0.75 AND no contradictions
    # Medium: score >= 0.60 AND >= 3 positive attribute hits
    # Low:    everything else
    attribute_hits = len(evidence)
    if final_score >= 0.75 and not contradictions:
        confidence = "\U0001f7e2 High"
    elif final_score >= 0.60 and attribute_hits >= 3:
        confidence = "\U0001f7e1 Medium"
    else:
        confidence = "\U0001f534 Low"

    return {
        "semantic_score":   float(semantic_score),
        "final_score":      float(final_score),
        "confidence_label": confidence,
        "evidence":         evidence,
        "contradictions":   contradictions,
        "keyword_overlap":  keyword_overlap,
    }

def save_found_item(user_id: str,
                    user_description: str,
                    ai_result: dict,
                    location_found: str = "Unknown") -> str:
    """Persist a found-item record with one canonical schema."""
    normalized = normalize_ai_result(ai_result)
    embed_text = build_found_embed_text(normalized, user_description, location_found)

    record = {
        "id":                   str(uuid.uuid4())[:8],
        "user_id":              user_id,
        "user_description":     user_description,
        "ai_description":       normalized["ai_description"],
        "item_type":            normalized["item_type"],
        "category":             normalized["category"],
        "color":                normalized["color"],
        "material":             normalized["material"],
        "brand":                normalized["brand"],
        "visible_contents":     normalized["visible_contents"],
        "distinctive_features": normalized["distinctive_features"],
        "condition":            normalized["condition"],
        "attribute_confidence": normalized["attribute_confidence"],
        "search_keywords":      normalized["search_keywords"],
        "location_found":       location_found,
        "date_found":           datetime.now(UTC).isoformat(),
        "status":               "available",
        "embed_text":           embed_text,
        "embedding":            embed(embed_text),
        "raw_ai_result":        dict(ai_result or {}),
    }
    found_items_db.append(record)
    return record["id"]

# IMPROVEMENT 3: Stricter search_found_items
# Changes vs v1:
#   Default threshold: 0.42 -> 0.55
#   Hard semantic pre-filter: items with semantic_score < 0.30 are skipped entirely
#   Hard category pre-filter: category-contradicting items are skipped before scoring

def search_found_items(lost_description: str, top_k: int = 3, threshold: float = 0.55) -> list[dict]:
    """
    Hybrid search: semantic similarity + structured evidence scoring.

    Improvements over v1:
    - threshold raised from 0.42 to 0.55
    - semantic pre-filter (score < 0.30 skipped)
    - category-contradiction hard pre-filter
    """
    valid_items = [
        item for item in found_items_db
        if item.get("status") == "available" and item.get("embedding") is not None
    ]
    if not valid_items:
        return []

    parsed_lost = parse_lost_item_description(lost_description)
    query_text  = build_lost_embed_text(parsed_lost, lost_description)

    query_vec       = np.array(embed(query_text)).reshape(1, -1)
    item_vecs       = np.array([item["embedding"] for item in valid_items])
    semantic_scores = cosine_similarity(query_vec, item_vecs)[0]

    lost_cat = parsed_lost.get("category")

    ranked_results = []
    for item, semantic_score in zip(valid_items, semantic_scores):

        # Hard pre-filter 1: semantic similarity floor
        if semantic_score < 0.30:
            continue

        # Hard pre-filter 2: skip category-contradicting items entirely
        found_cat = item.get("category")
        if (
            (lost_cat, found_cat) in CATEGORY_CONTRADICTIONS or
            (found_cat, lost_cat) in CATEGORY_CONTRADICTIONS
        ):
            continue

        scored = score_candidate(parsed_lost, item, float(semantic_score))

        if scored["final_score"] < threshold:
            continue

        result = dict(item)
        result.update({
            "score":            scored["final_score"],
            "semantic_score":   scored["semantic_score"],
            "confidence_label": scored["confidence_label"],
            "evidence":         scored["evidence"],
            "contradictions":   scored["contradictions"],
            "keyword_overlap":  scored["keyword_overlap"],
            "lost_structured":  parsed_lost,
        })
        ranked_results.append(result)

    ranked_results.sort(key=lambda x: x["score"], reverse=True)
    return ranked_results[:top_k]

print("DB helpers defined (improved matching system v2).")
print()
print("  Matching improvements summary:")
print("  1. Embed texts: synonym expansion removed -> precise vectors")
print("  2. Semantic weight: 0.58 -> 0.50 (attributes matter more)")
print("  3. Brand mismatch penalty: 0.08 -> 0.22 (Samsung != Apple)")
print("  4. Item-type mismatch: 0.18 -> 0.28 (+0.15 surcharge for known")
print("     contradictory pairs e.g. wallet vs backpack)")
print("  5. Category contradiction penalty: 0.20 -> 0.35")
print("  6. Threshold: 0.42 -> 0.55")
print("  7. Semantic pre-filter: skip items with score < 0.30")
print("  8. Category hard pre-filter before scoring")
print("  9. High confidence: score >= 0.75 (was 0.72)")
print("  10. Medium confidence: score >= 0.60 + 3 hits (was 0.52 + 2)")
````

## ðŸ”Ž Section 7 â€” DEMO: Pipeline 2 â€” Searching for a Lost Item

Simulates users who **lost** items.  
The AI embeds their description, runs cosine-similarity search, retrieves top matches,
then generates a human-readable explanation and contact suggestion for each match.

````python

def run_lost_item_search(lost_user_id: str, lost_description: str, top_k: int = 3):
    """
    Full Pipeline 2 flow:
      1. Parse and store lost description
      2. Hybrid search against found_items_db
      3. For each match: generate grounded AI explanation
      4. Print formatted report
    """
    user_name = users_db.get(lost_user_id, {}).get("name", lost_user_id)
    parsed_lost = parse_lost_item_description(lost_description)
    embed_text = build_lost_embed_text(parsed_lost, lost_description)

    lost_query_record = {
        "id": str(uuid.uuid4())[:8],
        "user_id": lost_user_id,
        "description": lost_description,
        "parsed_query": parsed_lost,
        "embedding": embed(embed_text),
        "created_at": datetime.now(UTC).isoformat(),
    }
    lost_queries_db.append(lost_query_record)

    print("\n" + "â•"*65)
    print("  LOST ITEM SEARCH")
    print("â•"*65)
    print(f"  Searched by : {user_name}")
    print(f"  Description : {lost_description}")
    print(f"  Parsed Type : {parsed_lost.get('item_type', 'Unknown')}")
    print(f"  Parsed Cat. : {parsed_lost.get('category', 'Other')}")
    print(f"  Parsed Attr : color={parsed_lost.get('color', 'Unknown')} | material={parsed_lost.get('material', 'Unknown')} | brand={parsed_lost.get('brand', 'Unknown')}")
    print("â•"*65)

    matches = search_found_items(lost_description, top_k=top_k)

    if not matches:
        print("  âŒ No matches found above the ranking threshold.")
        print("     Tip: try adding color, material, brand, or a distinctive detail.")
        return

    print(f"  âœ… {len(matches)} potential match(es) found:\n")

    for rank, match in enumerate(matches, 1):
        finder_name  = users_db.get(match["user_id"], {}).get("name", match["user_id"])
        finder_email = users_db.get(match["user_id"], {}).get("email", "N/A")
        score_pct    = match["score"] * 100
        semantic_pct = match["semantic_score"] * 100

        print(f"  {'â”€'*60}")
        print(f"  MATCH #{rank}  â€”  Confidence: {match['confidence_label']}  ({score_pct:.1f}%)")
        print(f"  {'â”€'*60}")
        print(f"  Item ID    : {match['id']}")
        print(f"  Item Type  : {match.get('item_type', 'Unknown')}")
        print(f"  Category   : {match['category']}")
        print(f"  Color      : {match['color']}")
        print(f"  Material   : {match.get('material', 'Unknown')}")
        print(f"  Brand      : {match['brand']}")
        print(f"  Location   : {match['location_found']}")
        print(f"  Date Found : {match['date_found'][:10]}")
        print(f"  AI Desc    : {match['ai_description']}")
        print(f"  Features   : {', '.join(match.get('distinctive_features', [])) or 'None listed'}")
        print(f"  Contents   : {', '.join(match.get('visible_contents', [])) or 'None listed'}")
        print(f"  Semantic   : {semantic_pct:.1f}%")

        if match.get("evidence"):
            print("  Why ranked : " + " | ".join(match["evidence"][:4]))
        if match.get("contradictions"):
            print("  Cautions   : " + " | ".join(match["contradictions"][:3]))

        print(f"\n  â³ Generating AI match explanation ...")
        explanation = explain_match(lost_description, match, match["score"])
        wrapped = textwrap.fill(explanation, width=60, initial_indent="  ", subsequent_indent="  ")
        print(f"\n  ðŸ’¬ AI Explanation:\n{wrapped}")

        print(f"\n  ðŸ“¬ Contact the finder:")
        print(f"     Name  : {finder_name}")
        print(f"     Email : {finder_email}")
        print(f"     â†’ In the app, tap 'Message {finder_name}' to open private chat")

        matches_db.append({
            "id": str(uuid.uuid4())[:8],
            "lost_query_id": lost_query_record["id"],
            "found_item_id": match["id"],
            "similarity_score": match["score"],
            "semantic_score": match["semantic_score"],
            "status": "pending",   # pending | confirmed | rejected
        })

    print(f"\n  {'â•'*65}")
    print(f"  {len(matches)} match(es) saved to matches table.")

print("âœ… run_lost_item_search() defined.")
````

## ðŸ“Š Section 8 â€” Matches Dashboard

````python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

if not matches_db:
    print("No matches yet â€” run the search cells above first.")
else:
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle("Lost & Found AI â€” Matches Dashboard", fontsize=14, fontweight="bold")

    # â”€â”€ Left: similarity score distribution â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    scores = [m["similarity_score"] for m in matches_db]
    colors = ["#2ecc71" if s >= 0.70 else "#f39c12" if s >= 0.45 else "#e74c3c" for s in scores]
    axes[0].barh(
        [f"Match {i+1}" for i in range(len(scores))],
        [s * 100 for s in scores],
        color=colors, edgecolor="white"
    )
    axes[0].axvline(70, color="#2ecc71", linestyle="--", alpha=0.7, label="High (70%)")
    axes[0].axvline(45, color="#f39c12", linestyle="--", alpha=0.7, label="Medium (45%)")
    axes[0].set_xlabel("Similarity Score (%)")
    axes[0].set_title("Match Confidence Scores")
    axes[0].legend(fontsize=8)
    axes[0].set_xlim(0, 100)

    # â”€â”€ Right: category breakdown of found items â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€â”€
    cats = [item["category"] for item in found_items_db]
    unique_cats = list(set(cats))
    cat_counts  = [cats.count(c) for c in unique_cats]
    palette = ["#3498db", "#9b59b6", "#e67e22", "#1abc9c", "#e74c3c"]
    axes[1].pie(
        cat_counts,
        labels=unique_cats,
        colors=palette[:len(unique_cats)],
        autopct="%1.0f%%",
        startangle=140,
        textprops={"fontsize": 10}
    )
    axes[1].set_title("Found Items by Category")

    plt.tight_layout()
    plt.savefig("/kaggle/working/matches_dashboard.png", dpi=150, bbox_inches="tight")
    plt.show()
    print("\nðŸ“Š Dashboard saved to /kaggle/working/matches_dashboard.png")
````
